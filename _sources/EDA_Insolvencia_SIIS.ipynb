## 4. Configuración y carga de datos

Los archivos originales están nombrados como `BS <año>.xlsx`, `IS <año>.xlsx` y `CF <año>.xlsx`, uno por cada estado financiero y año del panel (2017-2024). La lógica de carga y depuración se mantiene sin cambios respecto a las etapas previas del proyecto; lo que se amplía a partir de aquí es exclusivamente la capa de visualización.


```python
import pandas as pd
import numpy as np
from scipy.stats import kruskal

pd.set_option('display.float_format', lambda x: f'{x:,.0f}')

DATA_DIR = "Proyecto1"   
YEARS = [2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
```

## 5. Depuración de duplicados ("empresas repetidas")

Diagnóstico: en los archivos BS e IS cada empresa aparece dos veces por cada fecha de corte (columna `Periodo`: `"Periodo Actual"` vs. `"Periodo Anterior"`, que es el dato comparativo del período previo, no una empresa distinta). Además, un pequeño número de NITs por año (entre 7 y ~28, según el año) presentan un **corte intermedio adicional** (ej. a septiembre) sumado al de diciembre, porque radicaron un reporte extra ese año.

**Regla de deduplicación:** nos quedamos con `"Periodo Actual"` y, dentro de un mismo NIT-año, priorizamos el corte a 31 de diciembre; si no existe, se usa el corte más reciente disponible.


Tenemos un chequeo de calidad de datos (data-quality check), no todavía un análisis financiero. Compara, para cada tipo de estado financiero y año, cuántas filas había antes y después de depurar, y cuántas empresas quedaron con corte al 31-dic vs. con otra fecha. Si `filas_tras_dedup` no coincide con `nit_unicos_crudo`, algo quedó mal depurado es la validación de que la regla funcionó.

BS, IS y CF cuadran exactamente en `filas_tras_dedup` para los 8 años, y solo 2022 y 2024 tienen 1 NIT con corte distinto a dic-31 cada uno. Panel final: **22,521 filas empresa-año, 4,202 NITs únicos**. Valida el paso (b) de la metodología ("filtrar a Periodo Actual y corte 31-dic") la depuración fue limpia, sin duplicados arrastrados que infl en artificialmente el panel.


```python
def col(df, needle):
    for c in df.columns:
        if needle in c:
            return c
    raise KeyError(needle)

def load_and_dedupe(tipo, year):
    path = f"{DATA_DIR}/{tipo} {year}.xlsx"
    df = pd.read_excel(path)
    df['Fecha de Corte'] = pd.to_datetime(df['Fecha de Corte'])
    if 'Periodo' in df.columns:
        df = df[df['Periodo'] == 'Periodo Actual'].copy()
    df['es_dic31'] = (df['Fecha de Corte'].dt.month == 12) & (df['Fecha de Corte'].dt.day == 31)
    df = df.sort_values(['NIT', 'es_dic31', 'Fecha de Corte'], ascending=[True, True, True])
    df = df.drop_duplicates(subset='NIT', keep='last')  # se queda con dic-31 si existe, si no con el más reciente
    df['anio'] = year
    return df

panels = {'BS': {}, 'IS': {}, 'CF': {}}
resumen = []
for tipo in ['BS', 'IS', 'CF']:
    for year in YEARS:
        crudo = pd.read_excel(f"{DATA_DIR}/{tipo} {year}.xlsx")
        limpio = load_and_dedupe(tipo, year)
        panels[tipo][year] = limpio
        resumen.append({
            'tipo': tipo, 'anio': year,
            'filas_crudas': len(crudo), 'nit_unicos_crudo': crudo['NIT'].nunique(),
            'filas_tras_dedup': len(limpio),
            'con_corte_dic31': int(limpio['es_dic31'].sum()),
            'con_corte_distinto': int((~limpio['es_dic31']).sum()),
        })

resumen_df = pd.DataFrame(resumen)
tabla_estilo(resumen_df, 'Depuración de duplicados por tipo de estado financiero y año', precision=0)
```

## 6. Construcción del panel consolidado y cálculo de variables

Cruzamos BS + IS + CF por NIT dentro de cada año (join interno: solo empresas que reportaron los tres estados ese año) y calculamos las variables financieras del análisis. Se amplía el set de ratios (apalancamiento, ROA, margen neto, razón corriente) no puede quedarse solo en EBITDA/capital de trabajo/FCL, necesitamos las cuatro dimensiones clásicas de un análisis de crédito: **liquidez, endeudamiento, rentabilidad y flujo de caja**.


```python
records = []
for year in YEARS:
    bsY, isY, cfY = panels['BS'][year], panels['IS'][year], panels['CF'][year]

    ciiu_c = col(bsY, 'Clasificación Industrial')
    bs = bsY[['NIT', 'Razón social de la sociedad', ciiu_c,
              'Departamento de la dirección del domicilio', 'Fecha de Corte',
              'Activos corrientes totales (CurrentAssets)',
              'Pasivos corrientes totales (CurrentLiabilities)',
              'Total de activos (Assets)', 'Total pasivos (Liabilities)',
              'Patrimonio total (Equity)']].copy()
    bs = bs.rename(columns={ciiu_c: 'CIIU'})

    is_ = isY[['NIT',
               'Ganancia (pérdida) por actividades de operación (ProfitLossFromOperatingActivities)',
               'Ingresos de actividades ordinarias (Revenue)',
               'Ganancia (pérdida) (ProfitLoss)']].copy()

    dya_c = col(cfY, 'AdjustmentsForDepreciationAndAmortisationExpense')
    cfo_c = col(cfY, 'CashFlowsFromUsedInOperatingActivities')
    capex_ppe_c = col(cfY, 'PurchaseOfPropertyPlantAndEquipmentClassifiedAsInvestingActivities')
    capex_int_c = col(cfY, 'PurchaseOfIntangibleAssetsClassifiedAsInvestingActivities')
    cf = cfY[['NIT', dya_c, cfo_c, capex_ppe_c, capex_int_c]].copy()
    cf = cf.rename(columns={dya_c: 'dya', cfo_c: 'cfo', capex_ppe_c: 'capex_ppe', capex_int_c: 'capex_int'})

    m = bs.merge(is_, on='NIT', how='inner').merge(cf, on='NIT', how='inner')
    m['anio'] = year

    m['capital_trabajo'] = (m['Activos corrientes totales (CurrentAssets)']
                             - m['Pasivos corrientes totales (CurrentLiabilities)'])

    m['ebitda'] = (m['Ganancia (pérdida) por actividades de operación (ProfitLossFromOperatingActivities)'].fillna(0)
                   + m['dya'].fillna(0))

    capex = m['capex_ppe'].fillna(0).abs() + m['capex_int'].fillna(0).abs()
    m['fcl'] = m['cfo'].fillna(0) - capex

    records.append(m)

panel = pd.concat(records, ignore_index=True)
print(f"Panel consolidado: {panel.shape[0]} registros empresa-año, {panel['NIT'].nunique()} NITs únicos")
tabla_estilo(panel.head(), 'Vista previa del panel consolidado')
```

### 6.1 Variables adicionales: tamaño relativo, márgenes y ratios clásicos de análisis de crédito

El panel mezcla, dentro de las empresas más grandes de Colombia, compañías gigantes con compañías bastante más chicas. Comparar niveles absolutos en COP hace que las medianas y los gráficos por sector queden dominados por las empresas más grandes, sin importar cómo le vaya proporcionalmente a una empresa mediana del panel. Por eso se agregan variables **relativas** (márgenes, ratios), el estándar en analítica financiera de paneles heterogéneos, y se completan las cuatro dimensiones de riesgo de crédito: liquidez (razón corriente), endeudamiento (apalancamiento), rentabilidad (ROA, margen neto, margen EBITDA) y flujo de caja (ya calculado en la sección 5).


```python
panel['tamano'] = pd.qcut(
    np.log(panel['Total de activos (Assets)'].clip(lower=1)), 3,
    labels=['Pequeña', 'Mediana', 'Grande']
)

ingresos = panel['Ingresos de actividades ordinarias (Revenue)'].replace(0, np.nan)
activos = panel['Total de activos (Assets)'].replace(0, np.nan)
pasivos_corr = panel['Pasivos corrientes totales (CurrentLiabilities)'].replace(0, np.nan)

panel['ebitda_margen'] = ((panel['ebitda'] / ingresos) * 100).replace([np.inf, -np.inf], np.nan)
panel['margen_neto'] = ((panel['Ganancia (pérdida) (ProfitLoss)'] / ingresos) * 100).replace([np.inf, -np.inf], np.nan)
panel['roa'] = ((panel['Ganancia (pérdida) (ProfitLoss)'] / activos) * 100).replace([np.inf, -np.inf], np.nan)
panel['apalancamiento'] = ((panel['Total pasivos (Liabilities)'] / activos) * 100).replace([np.inf, -np.inf], np.nan)
panel['razon_corriente'] = (panel['Activos corrientes totales (CurrentAssets)'] / pasivos_corr).replace([np.inf, -np.inf], np.nan)

tabla_estilo(panel['tamano'].value_counts().to_frame('conteo'), 'Distribución de empresas por tamaño')
```

Los terciles de tamaño quedan casi perfectamente balanceados (Pequeña 7,506 / Grande 7,506 / Mediana 7,505), como se espera de `pd.qcut`. Esto es bueno para comparar tasas entre grupos en la sección 12.2 sin que el desbalance de tamaño de muestra distorsione la lectura.


## 7. Valores faltantes

Chequeo de calidad de datos (missing-data analysis). Antes de mirar distribuciones hay que saber cuántos datos realmente tenemos por variable un faltante alto podría sesgar cualquier estadística o modelo posterior. Una barra horizontal simple es la elección correcta aquí: son pocas categorías (variables) y no hay una dimensión temporal que narrar.


```python
variables = ['ebitda', 'capital_trabajo', 'fcl', 'roa', 'apalancamiento', 'razon_corriente', 'margen_neto']
titulos = ['EBITDA', 'Capital de trabajo', 'FCL', 'ROA (%)', 'Apalancamiento (%)', 'Razón corriente', 'Margen neto (%)']

faltantes = panel[variables].isna().sum()
faltantes_pct = (faltantes / len(panel) * 100).round(2)

orden = faltantes_pct.sort_values().index
titulos_map = dict(zip(variables, titulos))

fig = px.bar(
    x=faltantes_pct[orden].values, y=[titulos_map[v] for v in orden], orientation='h',
    color=faltantes_pct[orden].values, color_continuous_scale=ESCALA_CONTINUA,
    labels={'x': '% del panel', 'y': ''},
    text=[f'{v:.2f}%' for v in faltantes_pct[orden].values],
)
fig.update_traces(textposition='outside', cliponaxis=False, marker_line_color=COLOR_ANCLA, marker_line_width=0.5)
fig.update_coloraxes(showscale=False)
fig.update_xaxes(range=[0, faltantes_pct.max() * 1.25])
fig_layout(fig, '% de valores faltantes por variable', height=380)
fig.update_layout(margin=dict(l=140, r=60))
fig.show()

tabla_estilo(pd.DataFrame({'faltantes': faltantes, '% del panel': faltantes_pct}), 'Valores faltantes por variable')
```

Barras cerca de cero indican que la variable está casi completa y no necesita imputación agresiva. Los ratios derivados (ROA, apalancamiento, razón corriente) pueden tener algunos faltantes adicionales frente a EBITDA/capital de trabajo/FCL porque dependen de denominadores (ingresos, pasivos corrientes) que a veces son cero o nulos manejables con eliminación puntual de esas filas.

> **Interpretación:** Jerarquía clara de faltantes: EBITDA y FCL 0%, ROA y apalancamiento 0.02%, capital de trabajo 0.28%, razón corriente 0.38%, y **margen neto 7.10%** (muy por encima del resto). Tiene sentido: margen neto es el único ratio que depende de ingresos operacionales, y hay empresas-año con ingresos reportados en cero. El faltante no es aleatorio (MNAR: ingresos=0) imputar la media distorsionaría justo a las empresas más atípicas, por eso excluir caso a caso (regla de la sección 6.1) es la decisión correcta.


### 7.1 Cerrando la decisión: regla de tratamiento de datos faltantes

**Regla adoptada:** los ratios con denominador cero o nulo (ingresos, pasivos corrientes) ya quedan como `NaN` desde la sección 5.1 no se imputan con 0, porque eso falsearía el ratio (un ratio con denominador 0 no es '0', es 'no calculable'). Para el resto de faltantes residuales se excluye la fila **solo del análisis de esa variable puntual** (lo que ya hace `.dropna()` en cada sección); nunca se elimina la fila del panel completo por un faltante en una sola columna. 

```python
print(f'Filas totales del panel: {len(panel)}')
for var in variables:
    n_falt = panel[var].isna().sum()
    pct_falt = panel[var].isna().mean() * 100
    print(f"  {titulos_map[var]:22s}: {n_falt:5d} faltantes ({pct_falt:5.2f}%) -> se excluyen solo del analisis de esta variable")
```

### 8.1 Asimetría y curtosis: cuantificando lo que ya se intuye en la tabla

**Tipo de análisis:** univariado. La tabla anterior deja ver, por la distancia entre media y mediana, que hay asimetría, pero no la cuantifica. En un panel financiero, a diferencia de datos categóricos (ej. de salud, donde el dato suele ser un diagnóstico discreto), el chequeo estándar de un EDA de variables continuas es medir explícitamente qué tan lejos está cada variable de un comportamiento simétrico (asimetría) y qué tan pesadas son sus colas frente a una normal (curtosis en exceso), porque de ahí sale la decisión de qué transformación aplicar antes de modelar.


```python
from scipy.stats import skew, kurtosis

forma = pd.DataFrame({
    'asimetria': panel[variables].apply(lambda s: skew(s.dropna())),
    'curtosis_exceso': panel[variables].apply(lambda s: kurtosis(s.dropna())),  # exceso sobre la normal (Fisher)
}, index=variables)
forma.index = [titulos_map[v] for v in forma.index]

tabla_estilo(forma, 'Asimetria y curtosis por variable', precision=2)
```

`|asimetría| > 1` es asimetría fuerte (esperable en EBITDA, capital de trabajo y FCL, dado lo visto en la sección 8); `curtosis_exceso > 0` indica colas más pesadas que una normal (leptocúrtica) típico en variables financieras de nivel, donde unas pocas empresas gigantes generan valores extremos. Los ratios acotados (ROA, apalancamiento, razón corriente, margen neto) deberían salir más moderados; si alguno no lo es, es candidato adicional a revisar en la sección de outliers.


> **Interpretación:** Aquí se cuantifica lo anterior: las variables de nivel (EBITDA, capital de trabajo, FCL) tienen asimetría moderada (4.4 a 37.6), pero los ratios (ROA, apalancamiento, razón corriente, margen neto) tienen **asimetría de 120 a 150 y curtosis de 15,000 a 22,000**. No es una cola larga normal, es un puñado de filas rompiendo la escala. Estos 4 ratios son justamente los features centrales del modelo de scoring (tabla de la metodología, paso c) si se entrenan tal cual, ese puñado de empresas con denominador ~0 va a dominar cualquier regresión logística o el cálculo de VIF (ver sección 13.1).


### 8.2 Prueba formal de normalidad (Jarque-Bera)

La asimetría y curtosis de 8.1 ya insinúan que ninguna variable es normal, pero son estadísticos descriptivos, no una prueba con hipótesis y p-valor propio. Se formaliza con **Jarque-Bera**, que combina asimetría y curtosis en un solo estadístico y es válido asintóticamente para muestras grandes — a diferencia de **Shapiro-Wilk**, cuyo resultado deja de ser confiable a partir de n > 5,000, mientras que aquí cada variable tiene más de 20,000 observaciones. Es la pieza que le faltaba a la decisión, tomada más adelante, de usar pruebas no paramétricas (Kruskal-Wallis, Brown-Forsythe, Spearman): no solo "se ve asimétrico", hay una prueba formal que respalda ese diagnóstico.


```python
from scipy.stats import jarque_bera

resultados_jb = {}
for var in variables:
    datos_var = panel[var].dropna()
    stat_jb, pval_jb = jarque_bera(datos_var)
    resultados_jb[var] = {'estadistico JB': stat_jb, 'p-valor': pval_jb, 'n': len(datos_var)}

jb_df = pd.DataFrame(resultados_jb).T
jb_df.index = [titulos_map[v] for v in variables]
tabla_estilo(jb_df, 'Prueba de normalidad de Jarque-Bera (H0: la variable sigue una distribucion normal)', precision=3)
```

Un p-valor menor a 0.05 rechaza la hipótesis nula de normalidad para esa variable. Dado lo visto en 8.1 (asimetría y curtosis muy por encima de lo esperable bajo normalidad), se espera rechazar H0 en las 7 variables, lo cual respalda formalmente — y no solo por inspección visual del histograma — la elección de pruebas no paramétricas en el resto del notebook en vez de t-test/ANOVA/Pearson clásicos.


> **Interpretación:** Jarque-Bera rechaza la normalidad en las 7 variables con p≈0 — los estadísticos JB van de ~574 millones (capital de trabajo) hasta ~470,000 millones (apalancamiento), muy por encima de cualquier valor crítico razonable. Confirma con una prueba formal lo que 8.1 ya insinuaba con asimetría y curtosis: ni las variables de nivel ni los ratios acotados siguen una distribución normal, así que usar Kruskal-Wallis, Brown-Forsythe y Spearman en vez de sus equivalentes paramétricos (ANOVA, Levene clásico, Pearson) no fue una elección conservadora de más, era la única defendible con estos datos.


## 9. Distribución de cada variable: histograma + ECDF

**Tipo de análisis:** univariado forma de la distribución. Se usa **signed-log** en el eje X (con signo, porque hay valores negativos) para poder ver la forma real de la distribución sin que los outliers extremos aplasten el gráfico, y se complementa cada histograma con su **ECDF** (función de distribución empírica acumulada).


```python
def signed_log(x):
    return np.sign(x) * np.log1p(np.abs(x))

from plotly.subplots import make_subplots

vars_nivel = ['ebitda', 'capital_trabajo', 'fcl']
tit_nivel = ['EBITDA', 'Capital de trabajo', 'Flujo de caja libre (FCL)']

fig = make_subplots(rows=2, cols=3, subplot_titles=[f'{t} — histograma' for t in tit_nivel] + [f'{t} — ECDF' for t in tit_nivel],
                     vertical_spacing=0.14)

for i, (var, tit) in enumerate(zip(vars_nivel, tit_nivel)):
    datos = signed_log(panel[var].dropna())
    fig.add_trace(go.Histogram(x=datos, nbinsx=60, marker_color=PALETA_CAT[i], name=tit, showlegend=False), row=1, col=i + 1)

    ecdf_x = np.sort(datos)
    ecdf_y = np.arange(1, len(ecdf_x) + 1) / len(ecdf_x)
    fig.add_trace(go.Scatter(x=ecdf_x, y=ecdf_y, mode='lines', line=dict(color=PALETA_CAT[i], width=2), showlegend=False), row=2, col=i + 1)
    fig.add_vline(x=0, line_dash='dash', line_color=GRIS_CONTEXTO, row=2, col=i + 1)

fig.update_xaxes(title_text='signed log(valor)')
fig.update_yaxes(title_text='frecuencia', row=1)
fig.update_yaxes(title_text='% acumulado de empresas', row=2)
fig_layout(fig, 'Distribución de las variables ancla: histograma (signed-log) + ECDF', height=650)
fig.show()
```

Aun en escala logarítmica las tres distribuciones muestran colas largas y un pico marcado cerca de cero muchas empresas con valores pequeños/moderados y relativamente pocas con valores muy grandes (positivos o negativos). En la ECDF, el punto donde la curva cruza x=0 es directamente el % de empresas con esa variable en negativo el FCL cruza más arriba que EBITDA, coherente con que casi la mitad de las empresas tiene flujo de caja libre negativo en un año dado (normal cuando hay alta inversión, pero también señal de estrés si se sostiene).

> **Interpretación:** En escala signed-log, EBITDA y capital de trabajo tienen media ~9.4-9.5 (población concentrada en valores positivos), mientras que FCL tiene media 1.51 mucho más cerca de cero. Confirma numéricamente algo que se ve después en la sección 11: FCL es la variable ancla más frágil de las tres, no EBITDA.

### 10.1 Cerrando la decisión: winsorización al 1%/99%

La sección 9 diagnostica que los outliers son, con alta probabilidad, empresas grandes legítimas y no errores de captura. Se aplica **winsorización al percentil 1 y 99** (recorte de los valores más extremos al valor de ese percentil, no eliminación de filas) sobre las tres variables de nivel. A diferencia de eliminar filas, la winsorización preserva el tamaño de la muestra y no sesga el panel quitando justamente a las empresas grandes que son parte legítima del universo de estudio.


```python
def winsorizar(s, lim_inf=0.01, lim_sup=0.99):
    lo, hi = s.quantile([lim_inf, lim_sup])
    return s.clip(lo, hi)

for var in vars_nivel:
    panel[f'{var}_wz'] = winsorizar(panel[var])

fig = make_subplots(rows=1, cols=3, subplot_titles=tit_nivel, horizontal_spacing=0.08)
for i, (var, tit) in enumerate(zip(vars_nivel, tit_nivel)):
    datos_plot = signed_log(panel[f'{var}_wz'])
    fig.add_trace(go.Box(y=datos_plot, name=tit, marker_color=COLOR_ALERTA, line_color=COLOR_ALERTA,
                          fillcolor=COLOR_ANCLA, opacity=0.85, showlegend=False), row=1, col=i + 1)
fig.update_annotations(font_size=13)
fig_layout(fig, 'Variables de nivel tras winsorizacion 1%-99% (comparar con la seccion 8)', height=420, width=1000)
fig.show()
```

Comparar el rango intercuartílico y la posición de los bigotes contra la sección 9: deberían verse casi idénticos, porque la winsorización solo recorta el 1% más extremo de cada cola, mientras que los outliers más severos desaparecen. Las versiones `_wz` quedan disponibles en el panel exportado junto a las variables originales (sección 21), para que la siguiente etapa del proyecto elija cuál usar según el modelo.


> **Interpretación:** Winsorizar al 1%/99% recorta las colas (EBITDA pasa de [-20.8, 23.1] a [-17.3, 19.3]; capital de trabajo de [-22.6, 22.6] a [-18.4, 19.5]; FCL de [-22.1, 23.1] a [-18.5, 18.7]) **sin mover las medias** (9.46→9.46, 9.35→9.35, 1.51→1.51). Es la confirmación empírica de que winsorizar es seguro: se pierden extremos, no información central — clave porque con el label de insolvencia con solo 14 positivos no se puede dar el lujo de descartar filas completas.

## 12. Señales de estrés financiero (% de empresas en terreno negativo)

## 13. Comparación por periodo: pre-pandemia / pandemia / post-pandemia

Univariado agrupado por categoría temporal, resume la evolución en 3 bloques en vez de año a año, más fácil de leer cuando lo que importa es el contraste entre "antes", "durante" y "después" del choque de la pandemia.


```python
def asignar_periodo(anio):
    if anio <= 2019:
        return 'Pre-pandemia (2017-2019)'
    elif anio <= 2021:
        return 'Pandemia (2020-2021)'
    else:
        return 'Post-pandemia (2022-2024)'

orden_periodos = ['Pre-pandemia (2017-2019)', 'Pandemia (2020-2021)', 'Post-pandemia (2022-2024)']
panel['periodo'] = pd.Categorical(panel['anio'].apply(asignar_periodo), categories=orden_periodos, ordered=True)

tabla_estilo(panel.groupby('periodo', observed=True)[vars_nivel].median(), 'Mediana de indicadores por periodo')
```

Compara la empresa "típica" (mediana) de cada bloque. Si capital de trabajo o EBITDA caen en el bloque de pandemia respecto al de pre-pandemia y se recuperan (o no) en post-pandemia, es la primera señal cuantitativa del efecto del choque económico sobre las empresas del panel.

### 13.1 Cascada EBITDA → CFO → FCL por periodo (waterfall nativo)

En vez de comparar solo niveles absolutos, se usa una **cascada (waterfall)** por cada periodo: parte de EBITDA, muestra el efecto puente del capital de trabajo/impuestos, y llega a CFO y finalmente a FCL tras el CapEx. Es el formato estándar de análisis financiero corporativo para explicar cómo una cifra de rentabilidad contable se convierte, paso a paso, en caja realmente disponible y es muy superior aquí a un gráfico de barras sueltas, porque conecta visualmente el punto de partida con el de llegada mostrando cada "puente" intermedio.


```python
def cascada_fig(sub, titulo):
    ebitda_m = sub['ebitda'].median()
    cfo_m = sub['cfo'].median()
    fcl_m = sub['fcl'].median()
    ajuste_ktrab = cfo_m - ebitda_m
    capex_m = cfo_m - fcl_m

    fig = go.Figure(go.Waterfall(
        orientation='v',
        measure=['absolute', 'relative', 'total', 'relative', 'total'],
        x=['EBITDA', 'Ajuste KT +\nimpuestos', 'CFO', 'CapEx', 'FCL'],
        y=[ebitda_m, ajuste_ktrab, 0, -capex_m, 0],
        connector=dict(line=dict(color=GRIS_CONTEXTO, width=1)),
        increasing=dict(marker=dict(color=COLOR_SUAVE)),
        decreasing=dict(marker=dict(color=COLOR_ALERTA)),
        totals=dict(marker=dict(color=COLOR_ANCLA)),
        texttemplate='%{y:,.3s}',
    ))
    fig_layout(fig, titulo, height=420, yaxis_title='COP')
    return fig

anio_cascada = panel['anio'].max()
fig = cascada_fig(panel[panel['anio'] == anio_cascada], f'De EBITDA a Flujo de Caja Libre — mediana del panel ({anio_cascada})')
fig.show()

for periodo in orden_periodos:
    fig = cascada_fig(panel[panel['periodo'] == periodo], f'De EBITDA a FCL — {periodo}')
    fig.show()
```

Las barras oscuras (`COLOR_ANCLA`) son los puntos ancla (EBITDA, CFO, FCL); las barras "flotantes" son los puentes entre ellas, rojo claro si suman caja, rojo intenso si la restan. Compara la **altura final de la barra FCL** entre los tres periodos, no la altura de cada barra suelta: si "Pandemia" cierra en un FCL más bajo (o negativo) que "Pre-pandemia" y "Post-pandemia" no recupera ese nivel, es evidencia de un choque estructural persistente. Si el puente de CapEx es sistemáticamente mucho más grande que el de capital de trabajo, la inversión en activos fijos no el manejo de capital de trabajo es el principal freno entre rentabilidad operativa y caja libre.

> **Interpretación:**

| Periodo | EBITDA | Ajuste KT+impuestos | CapEx | FCL |
|---|---|---|---|---|
| Pre-pandemia | 2,065,296 | -1,364,835 (66%) | -690,075 | ~10,387 |
| Pandemia | 2,396,939 | -1,488,609 (62%) | -688,727 | ~219,603 |
| Post-pandemia | 3,659,063 | -2,566,452 (70%) | -950,687 | ~141,924 |

El "ajuste por capital de trabajo + impuestos" es, en las tres etapas, **más grande que el CapEx** - es el verdadero cuello de botella entre utilidad operativa y caja, no la inversión en activos fijos. Y ese ajuste se come una proporción creciente del EBITDA (66% - 70%), lo que explica por qué el % de empresas con FCL negativo no mejora estructuralmente después de la pandemia aunque EBITDA sí mejora (ver celda siguiente).


```python
tasas_periodo = panel.groupby('periodo', observed=True)[['ebitda_neg', 'capital_trabajo_neg', 'fcl_neg']].mean() * 100
tasas_periodo.columns = ['% EBITDA negativo', '% Capital trabajo negativo', '% FCL negativo']

fig = go.Figure()
for i, c in enumerate(tasas_periodo.columns):
    fig.add_trace(go.Bar(x=tasas_periodo.index, y=tasas_periodo[c], name=c, marker_color=PALETA_CAT[i * 2]))
fig_layout(fig, '% de empresas con cada variable en negativo, por periodo', height=450,
           yaxis_title='% de empresas', barmode='group')
fig.show()

tabla_estilo(tasas_periodo.round(1), '% de empresas con cada variable en negativo, por periodo')
```

A diferencia de la sección 11 (serie de 8 años, mejor como línea), aquí sí es correcto usar **barras**: son solo 3 categorías discretas (periodos) sin un eje temporal continuo que narrar entre ellas la comparación es "bloque contra bloque", no una tendencia. Si el % de empresas con EBITDA o capital de trabajo negativo sube claramente en "Pandemia", es evidencia directa de mayor riesgo de insolvencia en ese periodo.


> **Interpretación:** % EBITDA negativo mejora con el tiempo (20.4% - 20.9% - 18.4%), pero % FCL negativo no (45.3% - 41.4% - 44.1%, sin tendencia clara de mejora). Con una tasa base de FCL negativo cercana al 44%, esta variable sola es demasiado común para servir como señal aislada de riesgo frente a un evento de insolvencia que ocurre en <1% de los casos, necesita combinarse con otras señales, que es justo lo que hace el semáforo de la sección 17.


```python
fig = make_subplots(rows=3, cols=1, subplot_titles=[f'{t} por periodo (signed log)' for t in tit_nivel], vertical_spacing=0.1)
for i, (var, tit) in enumerate(zip(vars_nivel, tit_nivel)):
    datos_plot = panel[['periodo', var]].copy()
    datos_plot[var] = signed_log(datos_plot[var])
    for periodo in orden_periodos:
        # Un solo color para todas las cajas (el periodo ya está identificado en el eje x).
        sub = datos_plot.loc[datos_plot['periodo'] == periodo, var]
        fig.add_trace(go.Box(y=sub, name=periodo, marker_color=COLOR_ALERTA, line_color=COLOR_ALERTA,
                              fillcolor=COLOR_ANCLA, opacity=0.85, showlegend=False), row=i + 1, col=1)
    fig.update_yaxes(title_text='signed log(valor)', row=i + 1, col=1)

fig_layout(fig, 'Dispersión por periodo (boxplot)', height=900)
fig.show()
```

Mismo tipo de boxplot de la sección 9, pero agrupado por periodo en vez de por año, permite ver si la dispersión (no solo la mediana) de cada variable cambió durante la pandemia, por ejemplo si el rango intercuartílico se ensancha en "Pandemia" (mayor heterogeneidad entre empresas: algunas muy afectadas, otras no) y luego se contrae en "Post-pandemia".


**¿Es real la diferencia entre periodos, o podría ser azar?** Como las variables son muy asimétricas, una prueba t clásica no es apropiada. Usamos la prueba de **Kruskal-Wallis**, el equivalente no paramétrico de un ANOVA de un factor: compara las distribuciones de los 3 grupos sin asumir normalidad, usando rangos en vez de valores crudos.

```python
resultados_test = {}
for var in vars_nivel:
    grupos = [panel.loc[panel['periodo'] == p, var].dropna() for p in orden_periodos]
    stat, pval = kruskal(*grupos)
    resultados_test[var] = {'estadístico H': stat, 'p-valor': pval}

pd.DataFrame(resultados_test).T
```

Un **p-valor menor a 0.05** indica que la diferencia entre pre-pandemia, pandemia y post-pandemia es estadísticamente significativa (poco probable que se deba solo al azar) para esa variable. Un p-valor alto indicaría que, pese a lo que sugiere la gráfica, no hay evidencia estadística suficiente de que el periodo realmente influya en esa variable.

> **Interpretación:** Kruskal-Wallis confirma diferencias significativas entre periodos para las tres variables (H=203, 183, 52; p≈0 en las tres), y el boxplot de la celda anterior muestra que las medias en signed-log sí se mueven un poco entre periodos. Pero con n>22,000 casi cualquier diferencia es estadísticamente significativa — el tamaño real del desplazamiento es modesto, no hay que sobre-leer el p-valor como si la diferencia entre periodos fuera enorme (mismo punto se repite en la sección 17 con Cramér's V).


**Brown-Forsythe (homogeneidad de varianzas) y Dunn's post-hoc.** Kruskal-Wallis (celda anterior) solo dice que alguna mediana difiere entre periodos, no si la dispersión también cambia ni cuál par especifico es responsable. Brown-Forsythe (Levene centrado en la mediana, mas robusto que el Levene clásico ante colas pesadas) responde lo primero; Dunn's test con correccion de Bonferroni responde lo segundo.

```python
from scipy.stats import levene

resultados_bf = {}
for var in vars_nivel:
    grupos = [panel.loc[panel['periodo'] == p, var].dropna() for p in orden_periodos]
    stat, pval = levene(*grupos, center='median')  # Brown-Forsythe
    resultados_bf[var] = {'estadistico W': stat, 'p-valor': pval}

tabla_estilo(pd.DataFrame(resultados_bf).T, 'Brown-Forsythe: homogeneidad de varianzas entre periodos')
```

Con la varianza comparada, Dunn's identifica que par de periodos concentra la diferencia de mediana que ya detecto Kruskal-Wallis (requiere `pip install scikit-posthocs`):

```python
import scikit_posthocs as sp
from IPython.display import display

for var in vars_nivel:
    dunn = sp.posthoc_dunn(panel.dropna(subset=[var]), val_col=var, group_col='periodo', p_adjust='bonferroni')
    display(tabla_estilo(dunn.round(3), f'Dunn post-hoc (Bonferroni) - {var}: p-valor por par de periodos', precision=3))
```

#### Cerrando dos brechas: tamaño de efecto y corrección por comparaciones múltiples (Kruskal-Wallis)

La comparación por periodo reporta H y p-valor para las 3 variables de nivel, pero deja dos preguntas sin cerrar frente al mismo estándar que sí se aplicó más adelante: (1) para chi-cuadrado sí se calculó un tamaño de efecto (Cramér's V, sección 18) que evita sobre-leer el p-valor con n grande — a Kruskal-Wallis no se le había aplicado la misma disciplina; y (2) se corren 3 pruebas de Kruskal-Wallis independientes (una por variable) sin ajustar el umbral de significancia de forma conjunta, aunque Dunn's sí corrige con Bonferroni *dentro* de cada variable. Se completa con **eta-cuadrado basado en H** (tamaño de efecto estándar para Kruskal-Wallis) y con el ajuste de **Benjamini-Hochberg (FDR)** sobre los 3 p-valores.


```python
k_periodos = len(orden_periodos)

resultados_efecto = {}
for var in vars_nivel:
    n_var = panel[var].dropna().shape[0]
    H = resultados_test[var]['estadístico H']
    eta2_H = (H - k_periodos + 1) / (n_var - k_periodos)
    resultados_efecto[var] = {'H': H, 'n': n_var, 'eta2 (basado en H)': eta2_H}

tabla_estilo(pd.DataFrame(resultados_efecto).T, 'Tamano de efecto de Kruskal-Wallis (eta-cuadrado basado en H)', precision=4)
```

Se usa el mismo criterio orientativo que Cohen definió para eta-cuadrado clásico: ~0.01 efecto pequeño, ~0.06 mediano, ~0.14 o más grande. Permite responder si la diferencia entre periodos que ya confirmó el p-valor es grande o, como pasó con Cramér's V en la sección 18, estadísticamente real pero pequeña en magnitud.


```python
from statsmodels.stats.multitest import multipletests

pvals_kw = [resultados_test[var]['p-valor'] for var in vars_nivel]
rechaza_h0, pvals_ajustados, _, _ = multipletests(pvals_kw, alpha=0.05, method='fdr_bh')

correccion_multiple = pd.DataFrame({
    'p-valor original': pvals_kw,
    'p-valor ajustado (FDR-BH)': pvals_ajustados,
    'significativo tras ajuste (alfa=0.05)': rechaza_h0,
}, index=[titulos_map[v] for v in vars_nivel])
tabla_estilo(correccion_multiple, "Correccion de comparaciones multiples entre las 3 pruebas de Kruskal-Wallis (Benjamini-Hochberg / FDR)", precision=5)
```

El ajuste FDR-BH controla la tasa de falsos descubrimientos entre las 3 pruebas de Kruskal-Wallis, en vez de evaluar cada p-valor de forma aislada frente a 0.05. Con p-valores originales del orden de p≈0 en las tres variables, el ajuste no debería cambiar ninguna conclusión — pero dejar el chequeo explícito corrige una inconsistencia frente al mismo cuidado que sí se aplicó dentro de Dunn's.


> **Interpretación:** Los tres eta-cuadrado (EBITDA 0.0089, capital de trabajo 0.0081, FCL 0.0022) caen por debajo o justo en el umbral de 0.01 que Cohen asocia a un efecto "pequeño" — la misma historia que ya contó Cramér's V en la sección 18: el p-valor casi cero de Kruskal-Wallis (fiel reflejo de un n>22,000) no debe leerse como una diferencia grande entre periodos, sino como una diferencia real pero de magnitud modesta, y en FCL prácticamente nula. La corrección FDR-BH no cambia ninguna conclusión: los tres p-valores originales ya eran indistinguibles de 0 (redondean a 0.00000 con 5 decimales) y las tres variables siguen significativas tras el ajuste — el chequeo de comparaciones múltiples estaba pendiente, pero no había ningún resultado frágil escondido detrás de él.

**Tipo de análisis:** bivariado, mide qué tan relacionadas están las variables financieras entre sí. Se amplía el heatmap para incluir también los ratios de liquidez, endeudamiento y rentabilidad, no solo las tres variables de nivel.

Una correlación cercana a 1 indicaría redundancia; cercana a 0, independencia. EBITDA y FCL suelen mostrar una correlación moderada (el flujo de caja operativo parte de la utilidad operativa), mientras que capital de trabajo suele ser casi independiente de las otras dos, buena noticia para el modelo, porque cada variable aporta información distinta en vez de repetir la misma señal. Presta atención a si algún ratio (ej. ROA y margen neto) está altamente correlacionado, en ese caso, uno de los dos podría descartarse en la etapa de selección de variables sin perder poder predictivo.

> **Interpretación:** La correlación promedio entre las 8 variables es baja (0.213, máximo fuera de diagonal bien por debajo de 1) sugiere variables razonablemente independientes entre sí. Pero ojo con el VIF de la siguiente celda: no cuadra a simple vista con esta lectura de "poca redundancia", y la razón es que el VIF está contaminado por outliers (ver siguiente celda).


**Spearman y correlacion winsorizada (contraste de robustez).** Pearson mide relacion lineal y es sensible a outliers; con curtosis de hasta 22,375 en los ratios (seccion 8.1), unas pocas filas con denominador cercano a cero pueden estar dominando cada coeficiente, igual que domina el VIF de la siguiente celda. Se corre Spearman (basado en rangos, no requiere winsorizar) y, como contraste adicional, Pearson sobre los ratios winsorizados al 1%/99% si las tres matrices coinciden, hay mas confianza en que la relacion es real y no un artefacto de un puñado de outliers.

```python
ratios_sin_wz = ['roa', 'apalancamiento', 'razon_corriente', 'margen_neto']
for var in ratios_sin_wz:
    panel[f'{var}_wz'] = winsorizar(panel[var])

vars_corr_wz = [f'{v}_wz' if f'{v}_wz' in panel.columns else v for v in vars_corr]

corr_spearman = panel[vars_corr].corr(method='spearman')
corr_pearson_wz = panel[vars_corr_wz].corr()
corr_pearson_wz.columns = corr_pearson_wz.index = vars_corr  # mismos nombres para comparar lado a lado

fig = make_subplots(rows=1, cols=2, subplot_titles=['Spearman (rangos)', 'Pearson (ratios winsorizados 1%-99%)'],
                     horizontal_spacing=0.15)
fig.add_trace(go.Heatmap(z=corr_spearman.values, x=list(corr_spearman.columns), y=list(corr_spearman.index),
                          zmin=-1, zmax=1, colorscale=ESCALA_CONTINUA, showscale=False), row=1, col=1)
fig.add_trace(go.Heatmap(z=corr_pearson_wz.values, x=list(corr_pearson_wz.columns), y=list(corr_pearson_wz.index),
                          zmin=-1, zmax=1, colorscale=ESCALA_CONTINUA, showscale=True), row=1, col=2)
fig.update_annotations(font_size=13)
fig_layout(fig, 'Spearman vs. Pearson sobre ratios winsorizados (contraste frente a Pearson crudo)', height=520, width=1000)
fig.show()

diff = (corr_spearman - corr).abs()
tabla_estilo(diff.round(3), 'Diferencia |Spearman - Pearson crudo| (mayor = mas distorsion por outliers)', precision=2)
```

### 14.1 Multicolinealidad conjunta (VIF)

**Por qué no basta la matriz de correlación:** la correlación de la sección 13 es *pairwise* dos variables pueden no correlacionar fuerte entre sí y aun así compartir varianza con una tercera (ej. apalancamiento y razón corriente, ambas ligadas a la estructura de capital de la empresa). El **Variance Inflation Factor (VIF)** mide, para cada variable candidata, cuánta de su varianza ya está explicada por las demás en conjunto. Es el chequeo estándar de multicolinealidad conjunta en un panel financiero de variables continuas, y complementa el heatmap anterior.


```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

vars_vif = ['ebitda', 'capital_trabajo', 'fcl', 'roa', 'apalancamiento', 'razon_corriente', 'margen_neto', 'ebitda_margen']
datos_vif = panel[vars_vif].dropna()
X = add_constant(datos_vif)

vif = pd.DataFrame({
    'variable': vars_vif,
    'VIF': [variance_inflation_factor(X.values, i + 1) for i in range(len(vars_vif))],  # +1 salta la constante
}).sort_values('VIF', ascending=False)

tabla_estilo(vif.set_index('variable'), 'Factor de inflacion de varianza (VIF)', precision=2)
```

VIF > 5 (y sobre todo > 10) señala una variable redundante frente al resto del set — candidata a eliminar del modelo o a combinar con otra en un índice compuesto.


> **Interpretación:** El VIF calculado sobre las variables **sin winsorizar** da apalancamiento y ROA con VIF ≈ 256,091 (casi idénticos entre sí) un VIF así no es multicolinealidad real, es contaminación por outliers extremos. El patrón "apalancamiento y ROA con VIF prácticamente idéntico" es la misma huella de la sección 7.1: son casi seguro las mismas filas con activos ~0 dominando ambos cálculos a la vez. margen_neto y ebitda_margen (VIF ≈ 176 cada una) sí tiene una lectura económica razonable comparten "ingresos" como denominador. **Acción concreta:** recalcular VIF sobre las versiones winsorizadas antes de usarlo para descartar variables redundantes.


### 14.2 VIF recalculado sobre variables winsorizadas (cierre de la brecha señalada en 14.1)

La interpretación de 14.1 deja pendiente exactamente esta prueba: repetir el VIF sobre las mismas 8 variables, pero winsorizadas al 1%/99%, para separar multicolinealidad real de contaminación por outliers de denominador-casi-cero. `ebitda_wz`, `capital_trabajo_wz`, `fcl_wz` (sección 9.1) y `roa_wz`, `apalancamiento_wz`, `razon_corriente_wz`, `margen_neto_wz` (sección 14, contraste de robustez) ya existen en el panel; `ebitda_margen` no había sido winsorizado porque no participaba en esas secciones, así que se agrega aquí con el mismo criterio.


```python
panel['ebitda_margen_wz'] = winsorizar(panel['ebitda_margen'])

vars_vif_wz = [f'{v}_wz' for v in vars_vif]
datos_vif_wz = panel[vars_vif_wz].dropna()
X_wz = add_constant(datos_vif_wz)

vif_wz = pd.DataFrame({
    'variable': vars_vif,
    'VIF (winsorizado)': [variance_inflation_factor(X_wz.values, i + 1) for i in range(len(vars_vif_wz))],
})

comparacion_vif = vif.set_index('variable')[['VIF']].rename(columns={'VIF': 'VIF (sin winsorizar)'}) \
    .join(vif_wz.set_index('variable')) \
    .sort_values('VIF (winsorizado)', ascending=False)

tabla_estilo(comparacion_vif, 'VIF antes y despues de winsorizar (misma variable, mismo orden que en 14.1)', precision=2)
```

Si el VIF winsorizado de apalancamiento y ROA cae de ≈256,000 a un rango de una o dos cifras, confirma que el problema de 14.1 era enteramente contaminación por outliers y no multicolinealidad real entre las variables candidatas al modelo. Si algún par sigue con VIF > 10 tras winsorizar, eso sí sería evidencia de multicolinealidad genuina que habría que resolver antes de modelar (eliminar una de las dos variables o combinarlas en un índice compuesto), y no un artefacto de escala.


> **Interpretación:** Winsorizar resuelve el problema por completo. Apalancamiento y ROA caen de ≈256,091 a 1.27 y 1.55 respectivamente, y margen_neto / ebitda_margen bajan de 176.42 a 3.59 y 3.55. Las 8 variables winsorizadas quedan con VIF por debajo de 5 (el más alto es margen_neto con 3.59) — confirma que el VIF de 14.1 era enteramente contaminación por outliers de denominador ~0 y no multicolinealidad real entre las variables candidatas: ninguna resulta redundante frente a las demás, y las 8 pueden entrar juntas a un modelo sin inflar artificialmente sus coeficientes.

### 16.1 Variables relativas al sector

La sección 15 muestra que la rentabilidad varía por sector, pero hasta ahora cada empresa se compara contra el panel completo, no contra su propio sector la comparación quedó identificada en la sección 5.1, pero no calculada. ## 18. Semáforo financiero: evolución por año + estadística de significancia

Resume en una sola imagen la proporción de empresas "sanas", en "alerta" (una señal negativa) o en "riesgo alto" (dos o más señales negativas a la vez). Se baja el detalle a **nivel de año** (no solo los 3 bloques pre/pandemia/post) en **área apilada al 100%**: un área apilada enfatiza mejor la evolución continua de la composición que una barra apilada, y a nivel de año se ve si el cambio es gradual o si hay un año puntual que rompe la tendencia, algo que el bloque de 3 periodos, por diseño, no puede mostrar.

Un enfoque de *data analyst* no se queda en "se ve distinto": cierra la pregunta con estadística: **¿la composición del semáforo cambia de forma real entre periodos, o cabe dentro del ruido muestral?** Se responde con tres piezas complementarias: un test de independencia (chi-cuadrado), un tamaño de efecto (Cramér's V, porque con 22,521 observaciones hasta una diferencia mínima puede salir "significativa") y un intervalo de confianza por periodo (Wilson 95%) para saber si las diferencias punto a punto son distinguibles del azar.


```python
def clasificar_semaforo(row):
    negativos = sum([row['ebitda_neg'], row['fcl_neg']])
    if negativos == 0:
        return 'Sano'
    elif negativos == 1:
        return 'Alerta'
    else:
        return 'Riesgo alto'

panel['semaforo'] = panel.apply(clasificar_semaforo, axis=1)
orden_semaforo = ['Sano', 'Alerta', 'Riesgo alto']

# --- 1) Evolucion por año (area apilada al 100%) ---
tab_anio = pd.crosstab(panel['anio'], panel['semaforo'], normalize='index')[orden_semaforo] * 100
tab_anio = tab_anio.reindex(sorted(panel['anio'].unique()))

fig = go.Figure()
colores_semaforo = {'Sano': COLOR_SUAVE, 'Alerta': PALETA_CAT[3], 'Riesgo alto': COLOR_ANCLA}
for estado in orden_semaforo:
    fig.add_trace(go.Scatter(
        x=tab_anio.index, y=tab_anio[estado], name=estado, mode='lines',
        stackgroup='uno', line=dict(width=0.5, color=colores_semaforo[estado]),
        fillcolor=colores_semaforo[estado],
    ))
fig.add_vrect(x0=2019.5, x1=2021.5, fillcolor=GRIS_CONTEXTO, opacity=0.15, line_width=0)
fig_layout(fig, 'Semáforo financiero por año (% de empresas, área apilada al 100%)', height=480,
           xaxis_title='Año', yaxis_title='% de empresas')
fig.update_yaxes(range=[0, 100])
fig.show()

tabla_estilo(tab_anio.round(1), 'Semáforo financiero por año (%)')

# --- 2) ¿La diferencia entre periodos es real o es ruido muestral? ---
# H0: el estado del semaforo es independiente del periodo.
from scipy.stats import chi2_contingency

tabla_conteo = pd.crosstab(panel['periodo'], panel['semaforo'])[orden_semaforo]
chi2, p_valor, gl, _ = chi2_contingency(tabla_conteo)

n_total = tabla_conteo.values.sum()
k_efecto = min(tabla_conteo.shape) - 1
cramers_v = np.sqrt(chi2 / (n_total * k_efecto))

print(f"Test de independencia chi-cuadrado (semaforo x periodo):")
print(f"  chi2 = {chi2:.1f}, grados de libertad = {gl}, p-valor = {p_valor:.4g}")
print(f"  Cramer's V (tamano del efecto, 0=nulo .. 1=perfecto) = {cramers_v:.3f}")

# --- 3) Intervalo de confianza (Wilson 95%) del % en Riesgo alto, por periodo ---
from statsmodels.stats.proportion import proportion_confint

conteo_riesgo = panel.groupby('periodo', observed=True)['semaforo'].apply(lambda s: (s == 'Riesgo alto').sum())
n_periodo = panel.groupby('periodo', observed=True)['semaforo'].size()

ic_riesgo = pd.DataFrame({
    '% Riesgo alto': (conteo_riesgo / n_periodo * 100),
    'IC 95% inferior': [proportion_confint(k_, n_, method='wilson')[0] * 100 for k_, n_ in zip(conteo_riesgo, n_periodo)],
    'IC 95% superior': [proportion_confint(k_, n_, method='wilson')[1] * 100 for k_, n_ in zip(conteo_riesgo, n_periodo)],
}).round(1)
tabla_estilo(ic_riesgo, 'Intervalo de confianza (Wilson 95%) del % de empresas en Riesgo alto, por periodo')
```

- **Área apilada por año:** si el bloque "Riesgo alto" (más oscuro) crece de un año a otro y no vuelve a bajar, es evidencia visual directa de una huella persistente y al verlo año a año (no solo en 3 bloques) se distingue un cambio gradual de un salto puntual en un solo año.
- **Chi-cuadrado:** un p-valor menor a 0.05 dice que el semáforo *sí* depende del periodo de forma estadísticamente significativa, no es una impresión visual, es una diferencia real en la composición de las empresas.
- **Cramér's V:** con un panel de 22,521 empresas-año, hasta diferencias muy pequeñas pueden salir "estadísticamente significativas" en el chi-cuadrado. Cramér's V corrige esa trampa: valores por debajo de ~0.1 indican un efecto débil (la asociación es real pero pequeña en magnitud), aunque el p-valor sea diminuto, así se evita sobre-interpretar un resultado significativo pero económicamente poco relevante.
- **Intervalos de confianza (Wilson):** si el intervalo de "Riesgo alto" de dos periodos se solapa, la diferencia puntual entre esos dos periodos podría no ser distinguible del ruido muestral, incluso si el chi-cuadrado global (que compara los 3 periodos a la vez) resultó significativo, son preguntas distintas y complementarias, no contradictorias.


> **Interpretación:** "Sano" sube de 46.6% (2017) a 50.9% (2024), con un pico curioso en 2020 (51.8%, año de pandemia) coherente con el hallazgo contraintuitivo de la sección 11 (FCL mejorando en 2020 por freno de CapEx). "Riesgo alto" también repunta en 2020 (14.5%, su máximo) antes de bajar a ~11-12% después. El test chi-cuadrado da p=3.73e-05 (significativo) pero Cramér's V=0.024 **estadísticamente detectable pero económicamente casi nulo**. Con 22,000+ observaciones, un efecto minúsculo ya es "significativo"; el tamaño del efecto es el que dice que la composición del semáforo no cambió mucho en el fondo.


## 18.1 Nota metodológica: independencia de las observaciones (panel empresa-año)

Las pruebas de Kruskal-Wallis (sección 13) y chi-cuadrado (sección 18) asumen observaciones independientes, pero el panel es empresa-año: la misma empresa (NIT) aparece hasta 8 veces (una por año), aportando información correlacionada consigo misma de un año a otro. Ninguna de las dos pruebas corrige por esto (ni con errores estándar agrupados/*clustered*, ni con un modelo de efectos mixtos), lo que infla el n efectivo y, con eso, la significancia estadística — no invalida el hallazgo cualitativo, pero sí exige leer los p-valores con más cautela de la que ya pedía el tamaño de muestra grande. Se documenta la magnitud del problema y se repiten ambas pruebas con **una sola fila por empresa** (una observación aleatoria por NIT) como chequeo de robustez: si la conclusión no cambia con observaciones genuinamente independientes, el resultado es más confiable.


```python
n_obs_por_empresa = panel.groupby('NIT', observed=True).size()

resumen_repeticion = pd.DataFrame({
    'Estructura del panel': {
        'empresas-año en el panel': len(panel),
        'empresas unicas (NIT)': panel['NIT'].nunique(),
        'observaciones promedio por empresa': n_obs_por_empresa.mean(),
        'maximo de observaciones para una misma empresa': n_obs_por_empresa.max(),
    }
})
tabla_estilo(resumen_repeticion, 'La misma empresa aparece varias veces: no son observaciones independientes', precision=2)
```

## 19. Consistencia contable (chequeo de calidad)
Validación de calidad de datos (no descripción financiera). El balance general debe cumplir siempre la identidad contable **Activos = Pasivos + Patrimonio**. Esta sección no mide riesgo de insolvencia mide si el dato en sí es confiable, algo que debería revisarse antes de interpretar cualquier otra gráfica del notebook.


```python
panel['activos_calc'] = panel['Total pasivos (Liabilities)'] + panel['Patrimonio total (Equity)']
panel['error_contable'] = panel['Total de activos (Assets)'] - panel['activos_calc']
panel['error_contable_pct'] = (panel['error_contable'] / panel['Total de activos (Assets)'].replace(0, np.nan)) * 100

umbral = 1.0  # % de tolerancia
panel['inconsistente'] = panel['error_contable_pct'].abs() > umbral

# Diagnostico rapido: si la variable esta casi/perfectamente concentrada en 0 (identidad
# contable que se cumple por construccion en el reporte)
print(panel['error_contable_pct'].describe())
print(f"valores nulos (NaN): {panel['error_contable_pct'].isna().sum()} de {len(panel)}")

fig = px.histogram(
    panel[panel['error_contable_pct'].abs() < 20], x='error_contable_pct',
    color_discrete_sequence=[COLOR_ANCLA],
)
fig.add_vline(x=umbral, line_dash='dash', line_color=GRIS_CONTEXTO)
fig.add_vline(x=-umbral, line_dash='dash', line_color=GRIS_CONTEXTO)
fig_layout(fig, 'Error contable: Activos reportados vs. (Pasivos + Patrimonio), % del activo', height=420,
           xaxis_title='Error (%)', yaxis_title='Empresas-año')

rango_hist = max(panel['error_contable_pct'].abs().quantile(0.995), umbral) * 1.2
ancho_bin = rango_hist / 40
fig.update_traces(xbins=dict(start=-rango_hist, end=rango_hist, size=ancho_bin))
fig.update_xaxes(range=[-rango_hist, rango_hist])
fig.show()

n_inc = panel['inconsistente'].sum()
print(f"Empresas-año con inconsistencia contable > {umbral}%: {n_inc} de {len(panel)} ({n_inc/len(panel)*100:.2f}%)")

tabla_estilo(
    panel[panel['inconsistente']].sort_values('error_contable_pct', key=abs, ascending=False)
    [['NIT', 'Razón social de la sociedad', 'anio', 'Total de activos (Assets)', 'activos_calc', 'error_contable_pct']]
    .head(15),
    'Mayores inconsistencias contables (Activos != Pasivos + Patrimonio)'
)
```

La gran mayoría de los registros debería concentrarse muy cerca de 0% (la identidad contable se cumple, como es de esperar en estados financieros auditados/reportados a Supersociedades). Un error mayor al umbral no necesariamente es un error de captura, también puede deberse a partidas de "otros resultados integrales" no incluidas en las columnas que cruzamos, pero vale la pena decidir explícitamente si esas filas se excluyen, se marcan, o se dejan igual antes de pasar a la etapa de modelado.

> **Interpretación:** `error_contable_pct` tiene media -2.7e-19 y std 0  es ruido de punto flotante, no error real. 0 de 22,517 registros con inconsistencia >1%. La identidad contable Activos = Pasivos + Patrimonio se cumple en el 100% de los casos. Buena noticia estructural: los outliers extremos de las secciones 7.1/13.1/15.1 no son errores de digitación en el balance, son ratios genuinamente distorsionados por denominadores casi cero, un problema de escala/tratamiento estadístico, no de calidad de la fuente.


## 20. Balance del panel: empresas que entran y salen

**Tipo de análisis:** diagnóstico de estructura del panel (panel data), no financiero. Un panel de empresas-año casi nunca es "balanceado" no todas las empresas reportan los 8 años. Aquí importa especialmente para insolvencia: una empresa que **sale** del panel pudo simplemente dejar de estar entre las más grandes, pero también pudo entrar en liquidación.

```python
años_por_nit = panel.groupby('NIT')['anio'].nunique()

fig = px.histogram(años_por_nit, nbins=8, color_discrete_sequence=[COLOR_ANCLA])
fig_layout(fig, 'Número de años que cada empresa aparece en el panel (de 8 posibles)', height=400,
           xaxis_title='Años presente', yaxis_title='N° de empresas (NIT)')
fig.update_layout(showlegend=False)
fig.show()

pct_completas = (años_por_nit == 8).mean() * 100
print(f"Empresas presentes los 8 años: {(años_por_nit == 8).sum()} de {años_por_nit.shape[0]} ({pct_completas:.1f}%)")
```

```python
nits_por_anio = {a: set(panel.loc[panel['anio'] == a, 'NIT']) for a in sorted(panel['anio'].unique())}
años_ordenados = sorted(nits_por_anio.keys())

filas = []
for i, a in enumerate(años_ordenados):
    activas = len(nits_por_anio[a])
    if i == 0:
        entradas = salidas = np.nan
    else:
        anterior = nits_por_anio[años_ordenados[i - 1]]
        entradas = len(nits_por_anio[a] - anterior)
        salidas = len(anterior - nits_por_anio[a])
    filas.append({'anio': a, 'empresas_activas': activas, 'entradas': entradas, 'salidas': salidas})

balance_panel = pd.DataFrame(filas)

fig = go.Figure()
fig.add_trace(go.Scatter(x=balance_panel['anio'], y=balance_panel['empresas_activas'], mode='lines+markers',
                          name='Empresas activas', line=dict(color=COLOR_ANCLA, width=3)))
fig.add_trace(go.Bar(x=balance_panel['anio'], y=balance_panel['salidas'], name='Salidas', marker_color=PALETA_CAT[0]))
fig.add_trace(go.Bar(x=balance_panel['anio'], y=balance_panel['entradas'], name='Entradas', marker_color=COLOR_SUAVE))
fig_layout(fig, 'Empresas activas, entradas y salidas del panel por año', height=460,
           yaxis_title='N° de empresas (NIT)', barmode='group')
fig.show()

tabla_estilo(balance_panel, 'Balance del panel por año')
```

Un número de "salidas" alto y estable año a año es normal (rotación fuera del top de empresas grandes); un pico puntual de salidas coincidiendo con 2020-2021 sería una señal indirecta de insolvencias/liquidaciones reales que vale la pena investigar cuando se cruce con la Base de Datos de Insolvencia. El histograma de años-presente, por su parte, le dice al proyecto qué tan "balanceado" es el panel relevante para decidir si el modelo final se entrena a nivel empresa-año (panel) o a nivel empresa (corte transversal con variables agregadas).

> **Interpretación:** Solo 38.0% de los NITs (1,598 de 4,202) están los 8 años hay bastante rotación. El panel crece en número de empresas activas (2,421 - 3,018), pero 2024 es el primer año con salidas > entradas (238 vs 142, salida neta de -96). Punto importante para la metodología: como este panel es "las empresas más grandes por activos" y no un censo fijo, **salir del panel no equivale a insolvencia** puede ser que la empresa se encogió y salió del corte de tamaño, se fusionó, o dejó de reportar. El proxy real que define la metodología (corte extraordinario fuera de diciembre).


## 21. Persistencia del riesgo: transición del semáforo año a año

Matriz de transición (cadena de Markov de primer orden) a diferencia de la sección 17 (que mira la distribución del semáforo agregada por periodo), aquí se analiza qué tan probable es que **una misma empresa** cambie de estado de un año al siguiente. Es la pregunta más cercana a "predicción de insolvencia": ¿el riesgo es persistente (una empresa en riesgo tiende a seguir en riesgo) o es volátil (entra y sale de riesgo sin patrón)?


```python
panel_ord = panel.sort_values(['NIT', 'anio'])[['NIT', 'anio', 'semaforo']].copy()
panel_ord['anio_sig'] = panel_ord['anio'] + 1

transiciones = panel_ord.merge(
    panel_ord, left_on=['NIT', 'anio_sig'], right_on=['NIT', 'anio'],
    suffixes=('_t', '_t1'),
)[['NIT', 'anio_t', 'semaforo_t', 'semaforo_t1']]

orden_estados = ['Sano', 'Alerta', 'Riesgo alto']
matriz = pd.crosstab(transiciones['semaforo_t'], transiciones['semaforo_t1'], normalize='index').reindex(
    index=orden_estados, columns=orden_estados,
) * 100

fig = px.imshow(
    matriz, text_auto='.1f', color_continuous_scale=ESCALA_CONTINUA,
    labels=dict(x='Estado año t+1', y='Estado año t', color='% de empresas'),
)
fig_layout(fig, 'Matriz de transición del semáforo financiero (año t → año t+1)', height=460)
fig.show()

tabla_estilo(matriz.round(1), 'Matriz de transición (%, por fila)')
```

Cada fila suma 100% y responde "si una empresa estaba en este estado, ¿a cuál pasó al año siguiente?". La diagonal (Sano→Sano, Alerta→Alerta, Riesgo alto→Riesgo alto) mide **persistencia**: una diagonal alta en "Riesgo alto" confirma que el riesgo financiero tiende a sostenerse en el tiempo en vez de ser ruido de un solo año, justo el supuesto que hace viable predecir insolvencia con datos financieros históricos. Si en cambio las empresas saltan de estado sin patrón, el problema de predicción es intrínsecamente más difícil y probablemente necesite más años de historia por empresa como variable.

> **Interpretación:** "Riesgo alto" no es un estado absorbente: poco más de la mitad de las empresas que están ahí en el año t salen de esa categoría al año siguiente (55.1% = 37.2% a Alerta + 17.9% a Sano), y casi 1 de cada 5 salta directo a "Sano". Esto respalda la idea de la metodología de usar **delta interanual** (cambio de trayectoria) en vez de solo el nivel del año, el semáforo de un solo año tiene bastante ruido de transición, y un proxy de riesgo construido sobre 1-2 años consecutivos en "Riesgo alto" sería más estable que uno construido sobre un único corte anual.

## 23. Empresas destacadas por etapa (pre-pandemia / pandemia / post-pandemia)

**Tipo de análisis:** ranking + comparación de grupos (top performers vs. resto). Aquí se buscan las empresas con mejor desempeño en cada una de las 3 etapas, y luego se caracteriza qué tienen en común, no para explicar por qué a cada empresa individualmente le fue bien (no se cuenta con información cualitativa para eso), sino para detectar patrones agregados respaldados por los datos: sector, tamaño, apalancamiento.

**Criterio de "mejor desempeño"** (compuesto, para no depender de una sola cifra que un solo año atípico o el tamaño de la empresa puedan inflar):
- ROA promedio del periodo
- Margen EBITDA promedio del periodo
- % de años del periodo con FCL positivo

Cada métrica se convierte en un percentil dentro de su periodo (0-100) y el **score compuesto** es el promedio de los tres percentiles, así ROA, margen y consistencia de caja pesan por igual, sin importar su escala original.


```python
agg_empresa_periodo = panel.groupby(['NIT', 'periodo'], observed=True).agg(
    razon_social=('Razón social de la sociedad', 'first'),
    ciiu_2d=('ciiu_2d', 'first'),
    tamano=('tamano', lambda s: s.mode().iat[0] if not s.mode().empty else np.nan),
    roa_prom=('roa', 'mean'),
    margen_ebitda_prom=('ebitda_margen', 'mean'),
    pct_fcl_positivo=('fcl_neg', lambda s: (~s).mean() * 100),
    apalancamiento_prom=('apalancamiento', 'mean'),
    n_anios=('anio', 'nunique'),
).reset_index()

# Mapeo de división CIIU (2 dígitos) sección/sector, según CIIU Rev. 4 A.C. (DANE)
RANGOS_SECCION_CIIU = [
    (1, 3, 'Agricultura, ganadería, caza y pesca'), (5, 9, 'Minas y canteras'),
    (10, 33, 'Manufactura'), (35, 35, 'Electricidad, gas y aire acondicionado'),
    (36, 39, 'Agua, alcantarillado y gestión de desechos'), (41, 43, 'Construcción'),
    (45, 47, 'Comercio al por mayor y al por menor'), (49, 53, 'Transporte y almacenamiento'),
    (55, 56, 'Alojamiento y servicios de comida'), (58, 63, 'Información y comunicaciones'),
    (64, 66, 'Actividades financieras y de seguros'), (68, 68, 'Actividades inmobiliarias'),
    (69, 75, 'Actividades profesionales, científicas y técnicas'),
    (77, 82, 'Servicios administrativos y de apoyo'),
    (84, 84, 'Administración pública y defensa'), (85, 85, 'Educación'),
    (86, 88, 'Salud humana y asistencia social'),
    (90, 93, 'Actividades artísticas y de entretenimiento'), (94, 96, 'Otras actividades de servicios'),
    (97, 98, 'Actividades de los hogares como empleadores'),
    (99, 99, 'Organizaciones y entidades extraterritoriales'),
]

def ciiu_a_sector(codigo):
    try:
        n = int(codigo)
    except (ValueError, TypeError):
        return 'Sin clasificar'
    for ini, fin, nombre in RANGOS_SECCION_CIIU:
        if ini <= n <= fin:
            return nombre
    return 'Sin clasificar'

agg_empresa_periodo['sector'] = agg_empresa_periodo['ciiu_2d'].apply(ciiu_a_sector)

# Exigimos al menos 2 años de datos dentro del periodo para que el promedio sea razonable
agg_empresa_periodo = agg_empresa_periodo[agg_empresa_periodo['n_anios'] >= 2].dropna(
    subset=['roa_prom', 'margen_ebitda_prom', 'pct_fcl_positivo']
)

for col in ['roa_prom', 'margen_ebitda_prom', 'pct_fcl_positivo']:
    agg_empresa_periodo[col + '_pct'] = agg_empresa_periodo.groupby('periodo', observed=True)[col].rank(pct=True) * 100

agg_empresa_periodo['score_compuesto'] = agg_empresa_periodo[
    ['roa_prom_pct', 'margen_ebitda_prom_pct', 'pct_fcl_positivo_pct']
].mean(axis=1)

top10_por_periodo = (
    agg_empresa_periodo.sort_values('score_compuesto', ascending=False)
    .groupby('periodo', observed=True, group_keys=False)
    .head(10)
)

for p in orden_periodos:
    tabla = top10_por_periodo[top10_por_periodo['periodo'] == p].sort_values('score_compuesto', ascending=False)
    display(tabla_estilo(
        tabla[['razon_social', 'ciiu_2d', 'sector', 'tamano', 'roa_prom', 'margen_ebitda_prom', 'pct_fcl_positivo', 'score_compuesto']]
        .rename(columns={'razon_social': 'Empresa', 'ciiu_2d': 'CIIU', 'sector': 'Sector', 'tamano': 'Tamaño',
                          'roa_prom': 'ROA prom. (%)', 'margen_ebitda_prom': 'Margen EBITDA prom. (%)',
                          'pct_fcl_positivo': '% años FCL+', 'score_compuesto': 'Score compuesto'}),
        f'Top 10 empresas — {p}'
    ))
```

Cada tabla es el top 10 de esa etapa según el score compuesto (0-100, percentil promedio). Compara los nombres entre las tres tablas: si varias empresas se repiten en las tres etapas, es una señal de resiliencia estructural (no solo suerte en un periodo); si el top cambia por completo entre etapas, el "buen desempeño" es más situacional/coyuntural.

Y si ciertos sectores aparecen desproporcionadamente en el top, es evidencia de que el sector económico, más que decisiones puntuales de cada empresa explica buena parte del buen desempeño observado.

> **Interpretación:** El grupo de "empresas destacadas" (score compuesto alto) tiene apalancamiento mediano de **30.7% contra 56.0%** del panel completo, y está sesgado hacia empresas Grandes (50% vs 25.5% en el panel completo; Medianas subrepresentadas: 16.7% vs 34.3%). ## Síntesis para la siguiente etapa (interpretación cruzada con la metodología)

1. **Winsorizar antes de todo lo demás.** Los ratios (ROA, apalancamiento, razón corriente, margen neto) tienen outliers de denominador-casi-cero que contaminan estadística descriptiva, VIF y hasta el histograma sector-relativo. Ya resuelto en 9.1 para las variables de nivel; falta aplicar el mismo criterio a los 4 ratios antes de VIF y modelado.
2. **Recalcular VIF sobre datos winsorizados** — el resultado actual (VIF > 250,000) no sirve para decidir qué variables descartar.
3. **FCL es la variable de mayor riesgo estructural**, no EBITDA: ~44% negativo de forma persistente, transversal a todos los tamaños de empresa. El "tamaño" predice bien el riesgo de EBITDA pero no el de caja.
4. **Usar delta interanual / persistencia, no solo nivel** — la matriz de transición muestra que "Riesgo alto" es un estado con alta rotación (>55% sale en un año), coherente con la recomendación de la metodología de mirar la trayectoria y no solo la foto de un año.
5. **Cuidado con conteos chicos** al leer el mapa por departamento (Sucre 100%) y sectores con pocas empresas (CIIU R9, n=20) — cualquier porcentaje ahí necesita ir acompañado de su n.
6. **Salida del panel ≠ insolvencia** — para el label real hay que usar el proxy de corte extraordinario que define la metodología, no la desaparición de la muestra de empresas más grandes.


## 25. Cierre de brechas metodológicas

Esta versión del notebook cierra las cinco brechas identificadas en la revisión metodológica sobre la versión anterior, sin modificar el análisis, las decisiones ni las interpretaciones ya existentes:
